```python

import pandas as pd
import matplotlib.pyplot as plt


prices = [10, 12, 12, 14, 15, 15, 15, 18, 20, 29]

data = pd.Series(prices, name="Flower Retail Price")

count = data.count()
total = data.sum()
mean = data.mean()
median = data.median()
modes = data.mode()

minimum = data.min()
maximum = data.max()
price_range = data.max() - data.min()

q1 = data.quantile(0.25)
q2 = data.quantile(0.50)
q3 = data.quantile(0.75)
iqr = q3 - q1


sample_variance = data.var()
sample_std = data.std()


population_variance = data.var(ddof=0)
population_std = data.std(ddof=0)


skewness = data.skew()
kurtosis = data.kurt()
coefficient_of_variation = (sample_std / mean) * 100

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

outliers = data[
    (data < lower_fence) | (data > upper_fence)
]


print("=" * 45)
print("BASIC STATISTICAL FINDINGS")
print("=" * 45)

print(f"Count: {count}")
print(f"Sum: {total}")
print(f"Mean: {mean:.2f}")
print(f"Median: {median:.2f}")
print(f"Mode(s): {modes.tolist()}")

print(f"Minimum: {minimum:.2f}")
print(f"Maximum: {maximum:.2f}")
print(f"Range: {price_range:.2f}")

print(f"Q1 (25th percentile): {q1:.2f}")
print(f"Q2 (50th percentile): {q2:.2f}")
print(f"Q3 (75th percentile): {q3:.2f}")
print(f"IQR: {iqr:.2f}")

print(f"Sample Variance: {sample_variance:.2f}")
print(f"Sample Standard Deviation: {sample_std:.2f}")
print(f"Population Variance: {population_variance:.2f}")
print(f"Population Standard Deviation: {population_std:.2f}")

print(f"Skewness: {skewness:.3f}")
print(f"Excess Kurtosis: {kurtosis:.3f}")
print(f"Coefficient of Variation: {coefficient_of_variation:.2f}%")

print(f"Lower Outlier Fence: {lower_fence:.2f}")
print(f"Upper Outlier Fence: {upper_fence:.2f}")
print(f"Outliers: {outliers.tolist()}")

print("\nPANDAS DESCRIPTIVE SUMMARY")
print(data.describe())


fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].boxplot(data, vert=False)
axes[0].set_title("Box Plot of Flower Retail Prices")
axes[0].set_xlabel("Price (₹)")
axes[0].set_yticks([1])
axes[0].set_yticklabels(["Retail Price"])
axes[0].grid(axis="x", alpha=0.3)


axes[1].hist(data, bins=8, edgecolor="black")
axes[1].axvline(
    mean, linestyle="--",
    label=f"Mean = ₹{mean:.2f}"
)
axes[1].axvline(
    median, linestyle="-.",
    label=f"Median = ₹{median:.2f}"
)
axes[1].set_title("Distribution of Flower Retail Prices")
axes[1].set_xlabel("Price (₹)")
axes[1].set_ylabel("Frequency")
axes[1].legend()
axes[1].grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
