```python
'''Addition rule for disjoint events

A die is rolled 10,000 times.

Define: A=getting 1, B= getting 6, check the events are joint or disjoint

Verify: P(A∪B)=P(A)+P(B)'''

import random
trials=10000
getting1=0
getting6=0
for i in range(trials):
     outcome = random.choice([1,2,3,4,5,6])
     if outcome == 1:
      getting1 +=1
     elif outcome == 6:
      getting6 +=1
p_ab1=(getting1+getting6)/trials
p_a=getting1/trials
p_b=getting6/trials


print('experimental probablity of getting 1:',p_a)
print('experimental probablity of getting 6:',p_b)
print('experimental P(A U B):',p_a+p_b)# adding the experimental values of event a and b
print('addition of both events together:',p_ab1 )#here we found the joint probablity of event a and b occuring
if (p_ab1==(p_a+p_b)):# if both the values are equal then we can conclude that the events are disjoint
    print('events are disjoint')
else:
    print('events are not disjoint')
```

```python
'''Addition rule for overlapping events

For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6}

Verify: P(A∪B)=P(A)+P(B)−P(A∩B) '''
S = {1, 2, 3, 4, 5, 6} #sample space
A = {2, 4, 6} #event A
B = {4, 5, 6} #event B
union = A | B
intersection = A & B
P_A = len(A) / len(S)
P_B = len(B) / len(S)
P_intersection = len(intersection) / len(S)
P_union = len(union) / len(S)
print("P(A) =", P_A)
print("P(B) =", P_B)
print("P(A ∩ B) =", P_intersection)
print("P(A ∪ B) =", P_union)

print("\n To Verify the rule:")
print("P(A) + P(B) - P(A ∩ B) =",
      P_A + P_B - P_intersection)

print("P(A ∪ B) =", P_union)
```

```python
'''Multiplication rule for independent events

A coin is tossed and a die is rolled 10,000 times.

Define:

obtaining Head

obtaining an even number

Verify: P(A∩B)=P(A)P(B) '''

import random

n = 10000

head = 0
even = 0
head_even = 0

for i in range(n):
    coin = random.choice(["H", "T"])
    die = random.randint(1, 6)

    if coin == "H":
        head += 1

    if die % 2 == 0:
        even += 1

    if coin == "H" and die % 2 == 0:
        head_even += 1

# Experimental probabilities
P_A = head / n
P_B = even / n
P_AB_experimental = head_even / n # Corrected variable name

print("P(A) =", P_A)
print("P(B) =", P_B)
print("\n To Verify the rule:")
print("P(A ∩ B) =", P_AB_experimental)
print("P(A) × P(B) =", P_A * P_B)
```
