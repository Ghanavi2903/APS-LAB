```python
import matplotlib.pyplot as plt
import numpy as np
import numpy as np

p = 0.3
sample_size = 20

samples = np.random.binomial(
    n=1,
    p=p,
    size=sample_size
)

print(samples)
number_of_clicks = np.sum(samples == 1)

number_of_non_clicks = np.sum(samples == 0)

print("Number of clicks:", number_of_clicks)
print("Number of non-clicks:", number_of_non_clicks)
# Theoretical probabilities
theoretical = [0.70, 0.30]

# Empirical probabilities from the 20 observations
empirical = [
    number_of_non_clicks / sample_size,
    number_of_clicks / sample_size
]

x = np.array([0, 1])
width = 0.35

plt.bar(x - width/2, theoretical, width, label="Theoretical")
plt.bar(x + width/2, empirical, width, label="Empirical")

plt.xlabel("X")
plt.ylabel("Probability")
plt.title("Theoretical vs Empirical PMF")

plt.xticks(x, ["0 (No Click)", "1 (Click)"])
plt.legend()

plt.show()

# Display the values
print("Theoretical PMF:")
print("P(X=0) =", theoretical[0])
print("P(X=1) =", theoretical[1])

print("\nEmpirical PMF:")
print("P(X=0) =", empirical[0])
print("P(X=1) =", empirical[1])
```

```python
#step8
theoretical_mean = p

print("Theoretical Expectation =", theoretical_mean)
```

```python
#step9
empirical_mean = np.mean(samples)

print("Empirical Expectation =", empirical_mean)
```

```python
#step10
theoretical_variance = p * (1 - p)

print("Theoretical Variance =", theoretical_variance)
```

```python
#step11
empirical_variance = np.var(samples)

print("Empirical Variance =", empirical_variance)
```

```python
#step12
import pandas as pd

theoretical_values = [
    0.70,
    0.30,
    theoretical_mean,
    theoretical_variance
]

empirical_values = [
    number_of_non_clicks / sample_size,
    number_of_clicks / sample_size,
    empirical_mean,
    empirical_variance
]

absolute_difference = np.abs(
    np.array(theoretical_values) -
    np.array(empirical_values)
)

result_table = pd.DataFrame({
    "Measure": [
        "P(X=0) - No Click",
        "P(X=1) - Click",
        "Expectation",
        "Variance"
    ],

    "Theoretical Value": theoretical_values,
    "Empirical Value": empirical_values,
    "Absolute Difference": absolute_difference
})

print(result_table)
```

```python
#step13
sample_sizes = [10, 100, 1000, 10000]

results = []

for n in sample_sizes:

    # Generate random observations
    samples_n = np.random.binomial(
        n=1,
        p=p,
        size=n
    )

    # Count clicks and non-clicks
    clicks = np.sum(samples_n == 1)
    non_clicks = np.sum(samples_n == 0)

    # Empirical values
    empirical_p0 = non_clicks / n
    empirical_p1 = clicks / n
    empirical_mean_n = np.mean(samples_n)
    empirical_variance_n = np.var(samples_n)

    results.append([
        n,
        empirical_p0,
        empirical_p1,
        empirical_mean_n,
        empirical_variance_n
    ])

# Create table
sample_size_table = pd.DataFrame(
    results,
    columns=[
        "Sample Size",
        "Empirical P(X=0)",
        "Empirical P(X=1)",
        "Empirical Expectation",
        "Empirical Variance"
    ]
)

print(sample_size_table)
```
